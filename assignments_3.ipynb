Q11.
Write code to add value labels (data annotations) on top of each bar in a bar
chart. Use the dataset:
categories = ['A', 'B', 'C', 'D'] values = [23, 45, 12, 67]

```python
categories = ['A', 'B', 'C', 'D'] 
values = [23, 45, 12, 67]
plt.figure(figsize = (10,6))
bars = plt.bar(categories, values, color = ('red', 'blue', 'pink','yellow'))
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f'{height}',
        ha = 'center',
        va = 'bottom',
        fontsize = 10,
        fontweight = 600 
    )
plt.tight_layout()
plt.title('Data Annotations', fontweight = 'bold')
plt.xlabel('Categories', fontweight = 'bold')
plt.ylabel('Values', fontweight = 'bold')
    
plt.show()
```

Q17.
Write code to merge two DataFrames df1 and df2 on the common column
'EmpID' using an inner join. Also show how to perform a left join. Explain the
difference between inner and left joins

```python
df1 = pd.DataFrame({
    'EmpID':[101,102,103,104,105],
    'Name':['Ram','Sita','Hari','Gita','Shyam'],
    'Department':['IT','HR','Finance','IT','Marketing']
})

df2 = pd.DataFrame({
    'EmpID':[101,103,104,106],
    'Salary':[50000,60000,55000,70000],
    'City':['Kathmandu','Pokhara','Lalitpur','Biratnagar']
})
```

```python
# df = df1.merge(df2, on ='EmpID')
# df
or 
df = df1.merge(df2, on ='EmpID', how = 'inner')  # In pandas, the default join type in merge() is inner join, which returns only the rows with matching values in both DataFrames
df
```

```python
df = df1.merge(df2, on ='EmpID', how = 'left')  # The data are merged based on df1 because it is the left DataFrame, and all rows from df1 are kept.
df
```

```python
df = df1.merge(df2, on ='EmpID', how = 'right')  # # The data are merged based on df2 because it is the right DataFrame, and all rows from df2 are kept.
df
```

Q18.
Write Pandas code to:
(a) Use groupby() to find average marks per department.
(b) Apply pivot_table() to show average salary by gender and department.
(c) Use apply() to add a new column 'Grade' based on Score (A: >=90, B:
80–89, C: <80)

```python
data = {
    'Name':['Ram','Sita','Hari','Gita','Shyam','Rita','Aman','Soni'],
    'Department':['IT','HR','IT','Finance','HR','Finance','IT','HR'],
    'Gender':['M','F','M','F','M','F','M','F'],
    'Marks':[92,85,78,88,95,67,81,73],
    'Salary':[50000,60000,55000,65000,52000,70000,48000,62000]
}

df = pd.DataFrame(data)
df
```

```python
df.groupby('Department')['Marks'].mean().reset_index()
```

```python
df.pivot_table(values = 'Salary', index = 'Department' , columns = 'Gender')
```

```python
df['Grade'] = df['Marks'].apply(lambda x: 'A' if x >=90  else 'B' if 80<=x<90 else 'C')
df
```

Q19.
Case Study: You have a dataset 'students.csv' with columns: [Name, Subject,
Marks, Gender, Section].
Write complete Python code to:
(a) Load and explore the dataset (shape, dtypes, head). 
(b) Handle any missing values. 
(c) Create a bar chart showing average Marks per Subject. 
(d) Create a line plot showing Marks trend (sorted) for a specific student. 
(e) Create a grouped bar chart comparing average Marks by Gender for each
Subject.

```python
df = pd.read_csv('students.csv')
df.head()
```

```python
df.shape
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

```python
average = df.groupby('Subject')['Marks'].mean().round(2).reset_index()
average
```

```python
plt.figure(figsize= (10,6))
bars = plt.bar(average['Subject'], average['Marks'])
plt.title('Average marks as per subject', fontsize = 15, fontweight = 600)
plt.xlabel('Subjects',  fontsize = 11, fontweight = 60)
plt.ylabel('Average Marks',  fontsize = 11, fontweight = 60)
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f'{height}',
        ha = 'center',
        va = 'bottom',
        fontsize = 10,
        fontweight = 400)
    
plt.show()
```

```python
student = df[df['Name'] == 'Ram']
student
```

```python
plt.figure(figsize= (10,6))
bars = plt.bar(student['Subject'], student['Marks'])
plt.title(' marks as per subject', fontsize = 15, fontweight = 600)
plt.xlabel('Subjects',  fontsize = 11, fontweight = 60)
plt.ylabel(' Marks',  fontsize = 11, fontweight = 60)
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f'{height}',
        ha = 'center',
        va = 'bottom',
        fontsize = 10,
        fontweight = 400)
plt.grid(axis = 'y', linestyle = '--', alpha = 0.5, color = 'red')    
plt.show()
```

```python
avg = df.groupby(['Subject','Gender'])['Marks'].mean().round(2).reset_index()
pivot = avg.pivot(index='Subject', columns='Gender', values='Marks')
x = np.arange(len(pivot.index))
width = 0.35
plt.figure(figsize=(10,5))
plt.bar(x - width/2, pivot['F'], width, label='Female')
plt.bar(x + width/2, pivot['M'], width, label='Male')
plt.xticks(x, pivot.index)
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.title("Average Marks by Gender for Each Subject")
plt.legend()
plt.grid(axis='y')
plt.show()
```
